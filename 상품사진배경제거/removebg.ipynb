```python
import os  # 폴더 경로 처리를 위한 라이브러리
from rembg import remove  # 배경제거 함수 불러오기
from PIL import Image  # 이미지 열기/저장용 라이브러리
# ===== 입력 폴더 경로 =====
INPUT_DIR = r"C:\Users\User\OneDrive\문서\3yejoo\그라미스Gromise\상품구성\그라미스상품사진\누락된 사진들"  # 원본 이미지 폴더
# ===== 출력 폴더 경로 =====
OUTPUT_DIR = r"C:\Users\User\OneDrive\문서\3yejoo\그라미스Gromise\상품구성\그라미스상품사진\누락된 사진들_최종본"  # 결과 저장 폴더

os.makedirs(OUTPUT_DIR, exist_ok=True) # 출력 폴더 없으면 생성

file_list = os.listdir(INPUT_DIR) # 입력 폴더 파일 목록 가져오기

image_files = [ # 이미지 확장자만 필터링
f for f in file_list
if f.lower().endswith((".png", ".jpg", ".jpeg", ".webp"))
]

total_count = len(image_files) # 전체 개수 계산

if total_count == 0: # 처리할 파일 없으면 종료
    print("처리할 이미지가 없어") # 안내 출력
    raise SystemExit # 종료

current = 0 # 진행 카운트 초기화

for file_name in image_files: # 이미지 파일 반복
    current += 1 # 진행 카운트 증가
    input_path = os.path.join(INPUT_DIR, file_name) # 입력 경로 만들기
    percent = round((current / total_count) * 100, 1) # 퍼센트 계산
    print(f"[{current}/{total_count}] {percent}% 처리중 → {file_name}") # 진행 출력
    img = Image.open(input_path) # 이미지 열기
    result = remove(img) # 배경제거 실행 (RGBA 가능)
    base_name = os.path.splitext(file_name)[0] # 확장자 제거한 파일명만 추출
    output_file = f"{base_name}.png" # 출력 확장자는 무조건 png로 고정
    output_path = os.path.join(OUTPUT_DIR, output_file) # 출력 경로 만들기
    result.save(output_path) # PNG로 저장 (투명 유지)
print("모든 이미지 배경제거 완료") # 완료 메시지 출력 
```
