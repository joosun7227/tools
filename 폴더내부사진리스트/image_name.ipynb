```python
import os  # 파일/폴더 처리
import pandas as pd  # 엑셀 저장
import tkinter as tk  # GUI 기본
from tkinter import filedialog  # 파일 탐색기

# =========================
# tkinter 기본 설정
# =========================
root = tk.Tk()  # tkinter 객체 생성
root.withdraw()  # 빈 창 숨김

# =========================
# 폴더 선택
# =========================
IMG_DIR = filedialog.askdirectory(title="이미지 폴더 선택")  # 폴더 선택창

if not IMG_DIR:  # 선택 안 했을 경우
    raise SystemExit("폴더를 선택하지 않았어")  # 종료

# =========================
# 출력 파일 경로
# =========================
OUT_XLSX = os.path.join(IMG_DIR, "image_list.xlsx")  # 같은 폴더에 저장

# =========================
# 이미지 확장자
# =========================
IMG_EXTS = (".jpg", ".jpeg", ".png", ".webp", ".bmp")  # 이미지 확장자

# =========================
# 이미지 목록 수집
# =========================
rows = []  # 결과 저장 리스트

for fname in os.listdir(IMG_DIR):  # 폴더 안 파일 반복
    if fname.lower().endswith(IMG_EXTS):  # 이미지 파일만
        full_path = os.path.join(IMG_DIR, fname)  # 전체 경로
        ext = os.path.splitext(fname)[1]  # 확장자
        rows.append([fname, ext, full_path])  # 행 추가

# =========================
# 엑셀 저장
# =========================
df = pd.DataFrame(rows, columns=["파일명", "확장자", "전체경로"])  # 데이터프레임
df.to_excel(OUT_XLSX, index=False)  # 엑셀 저장

print("완료:", OUT_XLSX)  # 완료 출력
```
